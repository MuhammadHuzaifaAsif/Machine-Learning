```python
import numpy as np
x = np.array([2, 4, 4, 6, 9])
mean=np.mean(x)
median=np.median(x)
variance=np.var(x)
standard_deviation=np.std(x)
print(mean)
print(median)
print(variance)
print(standard_deviation)
```

```python
import pandas as pd

x = pd.Series([2, 4, 4, 6, 9])

mean = x.mean()
median = x.median()
mode = x.mode()
variance=x.var()
std=x.std()
print(mean)
print(median)
# print(mode)
print(variance)
print(std)
# to match euither this 0 or numpy 1
variance1=x.var(ddof=0)
print(pd.Series.var(x))
std1=x.std(ddof=0)
print(variance1)
print(std1)
print("Class call:", pd.Series.var(x,ddof=0))
```

```python
import numpy as np

x = np.array([2, 4, 4, 6, 9])

Q1 = np.percentile(x, 25)
Q3 = np.percentile(x, 75)

iqr = Q3 - Q1
lB = Q1 - (1.5 * iqr)
uB = Q3 - (1.5 * iqr)

print(Q1, Q3, iqr)
print(lB, uB)
```

```python
from sklearn.datasets import load_wine
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# sns=.set_theme()

wine = load_wine(as_frame=True)

df = wine.frame

print(df.head(10)) # dispalying first 10 rows of each column
print(df.shape) # shows number of rows and columsn
print(df.dtypes) # check dattypes for all coplumns
print(df.isnull().sum()) #check missing values
print(df.describe()) #gives mean std count(number of non missing values) quartiles  of each row min (smallest value) max(largest value)
print(df['alcohol'].value_counts())
df.hist(bins=50,density=True, color='g', alpha=0.75, figsize=(12,8))

Q1=df.quantile(0.25)
Q3=df.quantile(0.75)
iqr=Q3-Q1
lB=Q1-(1.5*iqr)
uB=Q3-(1.5*iqr)
summary=pd.DataFrame([lB,Q1,df.median(),Q3,uB])
plt.boxplot(summary,whis=[0,100])
plt.rotate(45)
des=df.describe()
plt.boxplot(des)
print(df.columns,df.dtypes)
plt.show()
corr = df.corr(method="spearman")

f, ax = plt.subplots(figsize=(9, 9))
sns.heatmap(corr, annot=True, fmt=".2f", linewidths=0.5, ax=ax)

print(df.var())
# skewness
print(df.skew())
print(df)
```
