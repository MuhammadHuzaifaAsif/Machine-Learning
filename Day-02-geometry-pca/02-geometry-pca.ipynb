```python
import numpy as np
x=np.array([2,3])
y=np.array([5,7])
# L1 distance function 

def my_l1_dis_func(x,y):
    return np.sum(np.abs(x-y))
# L2 distance function 
def my_l2_dis_func(x,y):
     return np.sqrt(np.sum(np.abs(x-y)**2))

# def cosine():
def cosine(x,y):
     dot=np.dot(x,y)
     l1=np.sqrt(np.sum(np.abs(x)**2))
     l2=np.sqrt(np.sum(np.abs(y)**2))
     return (dot/(l1*l2))



print(my_l1_dis_func(x,y))
print(my_l2_dis_func(x,y))
print(cosine(x,y))


```

```python
import numpy as np

np.random.seed(42)

X=np.random.rand(200,3)

# data center hoga
mean=np.mean(X,axis=0)
center=X-mean

# converiance matrix
convariance=np.cov(X,rowvar=False)

# eigen decompositon
values,vectors=np.linalg.eig(convariance)

# sort
sort=np.argsort(values)[::-1]
values=values[sort]
vectors=vectors[:,sort]
W=vectors[:,:2]

#PCA
z= center@W
print(z)
```

```python
np.random.seed(42)

X=np.random.rand(200,3)

# data center hoga
mean=np.mean(X,axis=0)
center=X-mean

# converiance matrix
convariance=np.cov(X,rowvar=False)

# eigen decompositon
values,vectors=np.linalg.eig(convariance)

# sort
sort=np.argsort(values)[::-1]
values=values[sort]
vectors=vectors[:,sort]
W=vectors[:,:2]

s=np.sum(W)

evr=W/s

print(np.round(evr*100,2))
```
