```python
import numpy as np

arr=np.array([2,3,4])
# list aur tuple to array mein badlt hai

arr2=np.array([2,3])
# is dimmension ka array output dega
# tamam zeroz hongae

arr3=np.ones((2,2))
# is dimension ka array banega
# tamam one honge

arr4=np.arange(0,10,2)
# zero sai shuru hoga
# 10 tak jaega
# 2 number skip karega

arr5=np.eye(3)
# yai ek identity matrix banaega
# 3x3

arr6=np.array([[4,8,3],[5,7,3]])
print(arr6.ndim)
# yai bata hai array mein kitn  rows kitni column hain

arr7=np.arange(6)
arr8=arr7.reshape(2,3)
# datako badla ga
# rowsxcol ki shakal mein

arr9=np.array([[1,2],[3,4],[8,7]])
print(arr9.T)
# transpose kardega


arr10=np.array([1,2,3])
arr11=np.array([5,6,7])
print(np.dot(arr10,arr11))
# dot product nikala ga


print(np.matmul(arr10,arr11))
# do matrices ka proper matrix multiplication karta hai
```

```python
import numpy as np

# create vectors
a=np.array([2,3,4])
b=np.array([5,6,7])

result=np.dot(a,b)

print(result)
```

```python
# create matrices

A=np.array([
    [1,2],
    [3,4]
])
B=np.array([
    [5,6],
    [7,8]
])

print("addition")
print(A+B)

print("multiplication")
print(np.matmul(A,B))

print("transpose")
print(A.T)

print("shape")
print(A.shape)

```
